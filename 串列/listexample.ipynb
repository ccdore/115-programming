# 簡介

串列是一個可以儲存多個元素(整數、浮點數、或字串)的容器

串列中每個元素都是使用逗點隔開，左右中有【】

```python
# score是一個串列
scores=[78,29,56,80,99]
print(scores)
```

## 索引值

我們可以使索引值(index)來存取串列中的元素

注意:索引值由零開始。

```python
print(scores[0])
print(scores[1])
```

## 串列內容的個數

len(a)取得a這個串列中元素的個數

len 是 length(長度) 的意思

```python
print(len(scores))
print(scores[len(scores)-1])
print(scores[-1])
```

## 新增元素

使用 `append()`新增元素到串列的尾端

```python
print(scores)
scores.append(78)
scores.append(93)
```

```python
print(scores)
scores[2]=70 #將70指定給索引值為2的元素
print(scores)
```

```python
weights=[12]*6
print(weights)
```

## 走訪串列的三種寫法


```python
print(scores)
for i in range(len(scores)):
    print(scores[i])
```

```python
#對於scores中的每個元素，一次取一個，放在e中
for e in scores:
    print(e)
```

```python
for i,e in enumerate(scores):
    print(i,e)
```
