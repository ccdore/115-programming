# Loop

Loop 指令可以讓程式碼依我的條件不斷地執行，直到滿足條件為止。

```python
# 底下的程式碼會印出0-9
for i in range(10):
    print(i)
```

```python
list(range(10))
```

```python
for i in range(10):
    print(i+1)
```

```python
sum=0
for i in range(10):
    sum = sum + (i + 1)

print(sum)
```

```python
print('&'*1)
print('&'*2)
print('&'*3)
print('&'*4)
print('&'*5)
print('&'*6)
```

```python
for i in range(6):
    print('&'*(i+1))
```

# range

python:
- 懶惰是一種美德
- 由0開始
- 包含頭，不包含尾

```python
print(list(range(5)))
# 區間，由0開始，到5-1結束
print(list(range(0,5)))
# 區間，由1開始，到5-1結束
print(list(range(1,5)))
# 區間，由1開始，到5-1結束，每次跳2個
print(list(range(1,5,2)))
```

```python
print(list(range(2,101,2)))
```
