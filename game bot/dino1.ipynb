```python
import pyautogui as pyg

pyg.sleep(2)  # sleep 2 sec.
pyg.press("space")  # 啟動

# 無窮迴圈
while True:
    im = pyg.screenshot()
    obs = False  # 假設沒有仙人掌

    for i in range(100):  # 檢查 100 個像素點
        c = im.getpixel((480 + i, 726))  # 使用 getpixel 並傳入座標 tuple
        if c == (83, 83, 83):
            obs = True
            break

    if obs == True:
        pyg.press("space")
```
